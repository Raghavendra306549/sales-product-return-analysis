```python
import pandas as pd
import os

# Load raw data
df = pd.read_csv("../data/raw/ecommerce_returns.csv")

# Parse dates (auto-detect format)
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Return_Date'] = pd.to_datetime(df['Return_Date'], errors='coerce')

# Handle missing values
df['Return_Reason'] = df['Return_Reason'].fillna('Not Returned')
df['Days_to_Return'] = df['Days_to_Return'].fillna(0)

# Target variable
df['Returned'] = df['Return_Status'].map({
    'Returned': 1,
    'Not Returned': 0
})

# Save cleaned data
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/cleaned_data.csv", index=False)

df.info()
df.head()
```
