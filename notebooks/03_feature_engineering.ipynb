```python
import pandas as pd

df = pd.read_csv("../data/processed/cleaned_data.csv")

df['Total_Order_Value'] = df['Product_Price'] * df['Order_Quantity']
df['High_Discount'] = (df['Discount_Applied'] > 40).astype(int)
df['Fast_Shipping'] = df['Shipping_Method'].isin(['Next-Day', 'Express']).astype(int)
df['Order_Month'] = pd.to_datetime(df['Order_Date']).dt.month

df = df.drop(columns=['Return_Status','Return_Date','Return_Reason'])

df.to_csv("../data/processed/final_data.csv", index=False)

print("Feature engineering completed")
```
